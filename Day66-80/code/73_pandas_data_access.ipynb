# Pandas DataFrame 数据访问

本节内容对应《深入浅出pandas-2》中的**索引和切片**以及**数据筛选**部分，重点介绍三种数据访问方式：

1. `loc` —— 基于标签（label）的数据访问
2. `iloc` —— 基于整数位置（integer position）的数据访问
3. 布尔索引（Boolean Indexing）—— 基于条件的数据筛选

## 0. 准备工作：创建示例 DataFrame

```python
import numpy as np
import pandas as pd

# 创建员工表 DataFrame（数据来自 hrs 数据库的 tb_emp 表）
emp_data = {
    'ename': ['胡一刀', '乔峰', '李莫愁', '张无忌', '丘处机',
              '欧阳锋', '张翠山', '黄蓉', '杨过', '朱九真',
              '苗人凤', '郭靖', '宋远桥', '张三丰'],
    'job': ['销售员', '分析师', '设计师', '程序员', '程序员',
            '程序员', '程序员', '销售主管', '会计', '会计',
            '销售员', '出纳', '会计师', '总裁'],
    'mgr': [3344, 7800, 2056, 2056, 2056, 3088, 2056, 7800,
            5566, 5566, 3344, 5566, 7800, np.nan],
    'sal': [1800, 5000, 3500, 3200, 3400, 3200, 4000, 3000,
            2200, 2500, 2500, 2000, 4000, 9000],
    'comm': [200, 1500, 800, np.nan, np.nan, np.nan, np.nan, 800,
             np.nan, np.nan, np.nan, np.nan, 1000, 1200],
    'dno': [30, 20, 20, 20, 20, 20, 20, 30, 10, 10, 30, 10, 10, 20]
}
emp_ids = [1359, 2056, 3088, 3211, 3233, 3244, 3251, 3344,
           3577, 3588, 4466, 5234, 5566, 7800]
emp_df = pd.DataFrame(data=emp_data, index=emp_ids)
emp_df.index.name = 'eno'
emp_df
```

## 1. 访问单列

获取 DataFrame 的某一列，可以用 **点号属性** 或 **方括号键名** 两种方式，返回结果是一个 `Series`。

```python
# 方式一：点号属性访问
emp_df.ename
```

```python
# 方式二：方括号键名访问
emp_df['job']
```

### 访问多列

使用花式索引（传入列名列表）获取多个列，返回结果仍然是一个 `DataFrame`。

## 2. `loc` —— 基于标签的数据访问

`loc` 使用**行标签和列标签**来定位数据，语法为 `df.loc[row_label, col_label]`。

> **重要特性**：当使用切片时，`loc` 的起止标签是**包含**两端的（闭区间），这与 Python 标准的切片行为不同。

### 2.1 用 `loc` 获取单行

```python
# 获取员工编号为 2056 的数据（返回 Series）
emp_df.loc[2056]
```

### 2.2 用 `loc` 获取多行（花式索引）

```python
# 传入标签列表获取多行（返回 DataFrame）
emp_df.loc[[2056, 7800, 3344]]
```

### 2.3 用 `loc` 切片获取连续多行

注意：`loc` 的切片是**闭区间**，即起止标签都会被包含。

```python
# 从 2056 到 3344 的所有行（包含 2056 和 3344）
emp_df.loc[2056:3344]
```

### 2.4 用 `loc` 同时指定行和列

推荐使用 `loc[label, col]` 的方式一次性指定行和列，避免链式索引。

```python
# 获取员工 2056 的职位
emp_df.loc[2056, 'job']
```

```python
# 获取多行多列
emp_df.loc[[2056, 7800], ['ename', 'job', 'sal']]
```

```python
# 行切片 + 列列表
emp_df.loc[2056:3344, ['ename', 'sal']]
```

### 2.5 用 `loc` 修改数据

```python
# 将员工 2056 的职位修改为'架构师'
emp_df.loc[2056, 'job'] = '架构师'
print('修改后员工 2056 的职位:', emp_df.loc[2056, 'job'])
```

```python
# 恢复原始数据
emp_df.loc[2056, 'job'] = '分析师'
```

## 3. `iloc` —— 基于整数位置的数据访问

`iloc` 使用**从 0 开始的整数位置**来定位数据，语法为 `df.iloc[row_pos, col_pos]`。

> **重要特性**：`iloc` 的切片行为与 Python 标准切片一致，是**左闭右开**区间。

### 3.1 用 `iloc` 获取单行

```python
# 获取第 2 行（索引位置 1，即员工 2056）
emp_df.iloc[1]
```

### 3.2 用 `iloc` 获取多行

```python
# 获取前 3 行
emp_df.iloc[:3]
```

```python
# 花式索引：获取第 0、2、4 行
emp_df.iloc[[0, 2, 4]]
```

### 3.3 用 `iloc` 同时指定行和列

```python
# 获取第 1 行第 1 列的值
print(f'第1行第1列: {emp_df.iloc[1, 1]}')

# 获取前 3 行的前 2 列
emp_df.iloc[:3, :2]
```

```python
# 行切片 + 列列表
emp_df.iloc[1:4, [0, 1, 3]]
```

### 3.4 `loc` 与 `iloc` 的对比

| 特性 | `loc` | `iloc` |
|------|-------|--------|
| 索引方式 | 标签（label） | 整数位置（position） |
| 切片区间 | **闭区间** `[start, end]` | **左闭右开** `[start, end)` |
| 适用场景 | 已知标签名 | 按位置顺序访问 |

```python
# 对比：loc 使用标签切片（闭区间，包含 3344）
print('loc[2056:3344] 行数:', len(emp_df.loc[2056:3344]))

# 对比：iloc 使用位置切片（左闭右开，不包含位置 7）
print('iloc[1:7] 行数:', len(emp_df.iloc[1:7]))
```

## 4. 布尔索引（Boolean Indexing）

通过布尔条件表达式生成一个布尔 Series，用它来筛选 DataFrame 的行。这是数据筛选最常用的方式之一。

### 4.1 单条件筛选

```python
# 筛选出月薪超过 3500 的员工
emp_df[emp_df.sal > 3500]
```

```python
# 查看布尔条件产生的 Series
mask = emp_df.sal > 3500
print(mask)
print(f'\n满足条件的员工数: {mask.sum()}')
```

### 4.2 多条件组合筛选

多个条件用 `&`（与）、`|`（或）、`~`（非）连接，**每个条件必须用括号括起来**。

```python
# 条件 1：月薪 > 3500 且 部门编号 == 20
emp_df[(emp_df.sal > 3500) & (emp_df.dno == 20)]
```

```python
# 条件 2：部门为 10 或 30 的员工
emp_df[(emp_df.dno == 10) | (emp_df.dno == 30)]
```

```python
# 条件 3：使用 isin() 判断是否在指定集合中
emp_df[emp_df.dno.isin([10, 30])]
```

### 4.3 布尔索引 + 列选择

可以在布尔筛选后继续选择特定的列。

```python
# 筛选月薪 > 3500 的员工，只看姓名、职位和月薪
high_salary = emp_df[emp_df.sal > 3500]
high_salary[['ename', 'job', 'sal']]
```

```python
# 也可以写成一行：先筛选行，再选列
emp_df.loc[emp_df.sal > 3500, ['ename', 'job', 'sal']]
```

### 4.4 处理缺失值的布尔筛选

```python
# 筛选出有补贴（comm 不为空）的员工
emp_df[emp_df.comm.notna()]
```

```python
# 筛选出没有补贴（comm 为空）的员工
emp_df[emp_df.comm.isna()]
```

### 4.5 使用 `query()` 方法筛选

`query()` 方法接受一个字符串表达式，语法更接近自然 Python 代码，可读性更好。

```python
# 等价于 emp_df[(emp_df.sal > 3500) & (emp_df.dno == 20)]
emp_df.query('sal > 3500 and dno == 20')
```

```python
# query 也支持变量引用，使用 @ 前缀
min_salary = 3500
dept_id = 20
emp_df.query('sal > @min_salary and dno == @dept_id')
```
