# Day 80 - Seaborn Statistical Visualization

Seaborn is a high-level data visualization library built on top of matplotlib. It provides:

- Dataset-oriented APIs for exploring relationships between multiple variables
- Support for categorical variables to display observations or summary statistics
- Options for visualizing univariate/bivariate distributions and comparing subsets
- Automatic estimation and plotting of linear regression models
- Integrated color palettes and themes for easy customization

This notebook covers: **seaborn setup**, **statistical plots**, **heatmaps**, **pair plots**, **C++ comparisons**, and **enterprise examples**.

## 1. Installation and Imports

````python
# Install seaborn if needed (uncomment the line below)
# %pip install seaborn
````

````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Set seaborn theme first
sns.set_theme()

# Then configure Chinese font support (must come AFTER set_theme)
plt.rcParams['font.sans-serif'].insert(0, 'SimHei')
plt.rcParams['axes.unicode_minus'] = False

print(f"Seaborn version: {sns.__version__}")
print(f"Pandas version: {pd.__version__}")
````

## 2. Loading the Tips Dataset

Seaborn ships with several built-in datasets. The `tips` dataset contains restaurant tipping data with columns:
- `total_bill`: total bill amount
- `tip`: tip amount
- `sex`: customer gender
- `smoker`: whether the customer smoked
- `day`: day of the week
- `time`: lunch or dinner
- `size`: party size

````python
import ssl
ssl._create_default_https_context = ssl._create_unverified_context

tips_df = sns.load_dataset('tips')
tips_df.info()
print()
tips_df.head(10)
````

## 3. Distribution Plots (histplot, kdeplot)

Understanding the distribution of a variable is the first step in exploratory data analysis. Seaborn's `histplot` combines a histogram with an optional KDE (kernel density estimate) curve.

**C++ comparison**: In C++, you would typically use a library like ROOT (CERN) or matplotlib-cpp, requiring significantly more boilerplate to produce the same result.

````python
# Histogram with KDE for total_bill distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(data=tips_df, x='total_bill', kde=True, ax=axes[0], color='steelblue')
axes[0].set_title('Total Bill Distribution (histplot + KDE)')
axes[0].set_xlabel('Total Bill ($)')

sns.kdeplot(data=tips_df, x='total_bill', hue='time', fill=True, ax=axes[1])
axes[1].set_title('Total Bill KDE by Meal Time')
axes[1].set_xlabel('Total Bill ($)')

plt.tight_layout()
plt.show()
````

## 4. Pair Plots - Exploring Variable Relationships

A pair plot shows pairwise relationships between all numeric columns in a dataset. The `hue` parameter color-codes points by a categorical variable, revealing patterns across groups.

This is one of seaborn's most powerful features -- a single line of code replaces what would be dozens of subplot calls in raw matplotlib or hundreds of lines in C++.

````python
# Pair plot colored by gender
sns.pairplot(data=tips_df, hue='sex', palette='Dark2')
plt.suptitle('Pair Plot of Tips Dataset (by Sex)', y=1.02)
plt.show()
````

````python
# Pair plot with different hue and specific columns
sns.pairplot(
    data=tips_df,
    vars=['total_bill', 'tip', 'size'],
    hue='smoker',
    palette='Set1',
    diag_kind='kde',
    plot_kws={'alpha': 0.6}
)
plt.suptitle('Pair Plot: Smoker vs Non-Smoker', y=1.02)
plt.show()
````

## 5. Joint Plots - Bivariate Analysis

Joint plots show the relationship between two variables along with their individual distributions on the margins.

````python
# Joint plot with regression line
g = sns.jointplot(
    data=tips_df, x='total_bill', y='tip',
    hue='sex', kind='scatter', height=7
)
g.set_axis_labels('Total Bill ($)', 'Tip ($)')
g.fig.suptitle('Joint Distribution: Bill vs Tip', y=1.02)
plt.show()
````

````python
# Verify the positive correlation between total_bill and tip
correlation = tips_df[['total_bill', 'tip']].corr()
print("Correlation matrix:")
print(correlation)
print(f"\nPearson correlation: {correlation.loc['total_bill', 'tip']:.4f}")
````

## 6. Regression Plots (lmplot)

Seaborn's `lmplot` automatically fits a linear regression model and plots the result with confidence intervals. This combines data visualization with statistical modeling in a single call.

````python
# Linear regression: total_bill vs tip, split by sex
sns.lmplot(
    data=tips_df, x='total_bill', y='tip',
    hue='sex', height=6, aspect=1.3,
    scatter_kws={'alpha': 0.5}
)
plt.title('Linear Regression: Bill vs Tip by Gender')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
````

````python
# Regression by day of the week
sns.lmplot(
    data=tips_df, x='total_bill', y='tip',
    col='day', col_wrap=2, height=4,
    scatter_kws={'alpha': 0.5}
)
plt.show()
````

## 7. Categorical Plots - Box Plots and Violin Plots

Box plots show the five-number summary (min, Q1, median, Q3, max) and outliers. Violin plots add kernel density estimation, showing the full shape of the distribution.

**C++ comparison**: Creating equivalent box plots in C++ with a library like Qt Charts or ROOT requires manually computing quartiles, setting up axes, and drawing each component. Seaborn does it all in one line.

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Box plot
sns.boxplot(data=tips_df, x='day', y='total_bill', ax=axes[0], palette='Set2')
axes[0].set_title('Box Plot: Total Bill by Day')
axes[0].set_xlabel('Day of Week')
axes[0].set_ylabel('Total Bill ($)')

# Violin plot
sns.violinplot(data=tips_df, x='day', y='total_bill', ax=axes[1], palette='Set2')
axes[1].set_title('Violin Plot: Total Bill by Day')
axes[1].set_xlabel('Day of Week')
axes[1].set_ylabel('Total Bill ($)')

plt.tight_layout()
plt.show()

print("Note: Violin plots show the full density distribution, not just quartiles.")
print("Box plots explicitly mark outliers as individual points.")
````

````python
# Swarm plot - shows every data point without overlap
plt.figure(figsize=(10, 6))
sns.swarmplot(data=tips_df, x='day', y='total_bill', hue='sex', palette='muted', size=4)
plt.title('Swarm Plot: Every Data Point by Day and Gender')
plt.xlabel('Day of Week')
plt.ylabel('Total Bill ($)')
plt.legend(title='Sex')
plt.show()
````

## 8. Heatmaps - Correlation Matrices

Heatmaps are essential for visualizing correlation matrices, confusion matrices, and any 2D grid data. They use color intensity to represent values, making patterns immediately visible.

**C++ comparison**: Generating a heatmap in C++ (e.g., with OpenCV or Qt) requires iterating over a 2D array, mapping values to colors, and rendering each cell. Seaborn's `heatmap()` handles annotations, color mapping, and axis labels with a single function call.

````python
# Correlation heatmap of numeric columns
numeric_cols = tips_df.select_dtypes(include=[np.number])
corr_matrix = numeric_cols.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    square=True,
    linewidths=0.5,
    cbar_kws={'shrink': 0.8}
)
plt.title('Correlation Heatmap of Tips Dataset')
plt.tight_layout()
plt.show()
````

````python
# Pivot table heatmap: average tip by day and time
pivot = tips_df.pivot_table(values='tip', index='day', columns='time', aggfunc='mean')

plt.figure(figsize=(8, 5))
sns.heatmap(
    pivot,
    annot=True,
    fmt='.2f',
    cmap='YlOrRd',
    linewidths=1,
    cbar_kws={'label': 'Average Tip ($)'}
)
plt.title('Average Tip by Day and Meal Time')
plt.xlabel('Meal Time')
plt.ylabel('Day of Week')
plt.tight_layout()
plt.show()
````

## 9. C++ Comparison: Why Seaborn Wins for Data Science

Below is a summary of the effort required to produce equivalent visualizations in C++ vs. Python/seaborn.

| Task | Python (Seaborn) | C++ (Qt Charts / ROOT / matplotlib-cpp) |
|------|-----------------|----------------------------------------|
| Install | `pip install seaborn` | Build from source or install binary SDKs |
| Histogram + KDE | `sns.histplot(data, x, kde=True)` | Manual bin calculation, KDE kernel, custom drawing |
| Pair plot | `sns.pairplot(data, hue)` | Nested loops, N*(N-1) subplots, manual axis setup |
| Heatmap | `sns.heatmap(corr)` | 2D array iteration, color mapping, annotation rendering |
| Regression line | `sns.lmplot(data, x, y)` | Implement least-squares, compute CI, draw line + band |
| Violin plot | `sns.violinplot(data, x, y)` | KDE computation per group, mirrored density drawing |
| Code lines (typical) | 1-5 lines | 50-200+ lines |

**Key takeaway**: Seaborn excels at exploratory data analysis (EDA) where speed of iteration matters. C++ libraries are better suited for high-performance rendering of final production charts in embedded systems or real-time dashboards.

## 10. Enterprise Example: Sales Performance Dashboard

In a real enterprise setting, you might analyze sales data across regions, product lines, and time periods. This example simulates a quarterly sales dataset and demonstrates multi-panel visualization.

````python
# Simulate enterprise sales data
np.random.seed(42)
n = 500

regions = ['North America', 'Europe', 'Asia Pacific', 'Latin America']
products = ['Cloud Services', 'Enterprise Licenses', 'Consulting', 'Support Contracts']
quarters = ['Q1', 'Q2', 'Q3', 'Q4']

sales_df = pd.DataFrame({
    'region': np.random.choice(regions, n),
    'product': np.random.choice(products, n),
    'quarter': np.random.choice(quarters, n),
    'revenue': np.random.lognormal(mean=10, sigma=0.8, size=n).round(2),
    'deal_size': np.random.randint(1, 50, n),
    'satisfaction': np.clip(np.random.normal(7.5, 1.2, n), 1, 10).round(1)
})

sales_df['revenue_millions'] = (sales_df['revenue'] / 1000).round(2)
print(f"Dataset: {len(sales_df)} sales records")
sales_df.head()
````

````python
# Enterprise dashboard: 4-panel visualization
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Panel 1: Revenue by region (box plot)
sns.boxplot(data=sales_df, x='region', y='revenue_millions', palette='Set2', ax=axes[0, 0])
axes[0, 0].set_title('Revenue Distribution by Region')
axes[0, 0].set_xlabel('')
axes[0, 0].set_ylabel('Revenue (millions $)')
axes[0, 0].tick_params(axis='x', rotation=20)

# Panel 2: Correlation heatmap
num_cols = sales_df[['revenue_millions', 'deal_size', 'satisfaction']]
sns.heatmap(num_cols.corr(), annot=True, fmt='.2f', cmap='RdYlGn',
            center=0, ax=axes[0, 1], square=True)
axes[0, 1].set_title('Metric Correlations')

# Panel 3: Revenue trend by quarter and region
quarterly = sales_df.groupby(['quarter', 'region'])['revenue_millions'].mean().reset_index()
sns.barplot(data=quarterly, x='quarter', y='revenue_millions', hue='region',
            palette='muted', ax=axes[1, 0])
axes[1, 0].set_title('Average Revenue by Quarter and Region')
axes[1, 0].set_ylabel('Avg Revenue (millions $)')
axes[1, 0].legend(title='Region', fontsize=8)

# Panel 4: Satisfaction vs Revenue scatter
sns.scatterplot(
    data=sales_df, x='satisfaction', y='revenue_millions',
    hue='product', size='deal_size', sizes=(20, 200),
    alpha=0.6, ax=axes[1, 1]
)
axes[1, 1].set_title('Satisfaction vs Revenue (size = deal count)')
axes[1, 1].set_xlabel('Customer Satisfaction')
axes[1, 1].set_ylabel('Revenue (millions $)')
axes[1, 1].legend(fontsize=7, loc='upper left')

plt.suptitle('Enterprise Sales Performance Dashboard', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
````

## 11. Enterprise Example: Customer Segmentation Heatmap

Heatmaps are commonly used in enterprise analytics to visualize segment-level metrics, enabling quick identification of high-value and at-risk customer groups.

````python
# Simulate customer segmentation data
np.random.seed(123)
segments = ['Enterprise', 'Mid-Market', 'SMB', 'Startup']
metrics = ['Retention %', 'Avg Deal Size', 'NPS Score', 'Expansion Rate %', 'Churn Risk %']

seg_data = pd.DataFrame(
    np.random.rand(len(segments), len(metrics)) * 100,
    index=segments,
    columns=metrics
).round(1)

# Adjust some values to be more realistic
seg_data.loc['Enterprise', 'Retention %'] = 95.2
seg_data.loc['Enterprise', 'NPS Score'] = 72.0
seg_data.loc['Startup', 'Churn Risk %'] = 38.5

plt.figure(figsize=(10, 5))
sns.heatmap(
    seg_data, annot=True, fmt='.1f', cmap='RdYlGn_r',
    linewidths=2, linecolor='white',
    cbar_kws={'label': 'Score'}
)
plt.title('Customer Segment Health Scorecard', fontsize=14, fontweight='bold')
plt.ylabel('Customer Segment')
plt.xlabel('Metric')
plt.tight_layout()
plt.show()

print("Interpretation: Higher values in Retention/NPS/Expansion are good.")
print("Higher values in Churn Risk are bad (note the reversed colormap).")
````

## 12. Enterprise Example: A/B Test Results Visualization

A/B testing is standard practice in product and marketing analytics. Seaborn makes it easy to compare distributions across test groups.

````python
# Simulate A/B test data for a checkout flow experiment
np.random.seed(77)
n_per_group = 300

ab_df = pd.DataFrame({
    'group': ['Control'] * n_per_group + ['Variant A'] * n_per_group + ['Variant B'] * n_per_group,
    'conversion_time_sec': np.concatenate([
        np.random.normal(120, 30, n_per_group),   # Control
        np.random.normal(95, 25, n_per_group),     # Variant A (faster)
        np.random.normal(105, 35, n_per_group)     # Variant B (slightly faster, more variance)
    ]),
    'purchase_amount': np.concatenate([
        np.random.lognormal(3.5, 0.5, n_per_group),
        np.random.lognormal(3.7, 0.5, n_per_group),
        np.random.lognormal(3.6, 0.6, n_per_group)
    ])
})

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Conversion time comparison
sns.violinplot(data=ab_df, x='group', y='conversion_time_sec',
               palette='pastel', inner='box', ax=axes[0])
axes[0].set_title('Checkout Conversion Time (A/B Test)')
axes[0].set_xlabel('Test Group')
axes[0].set_ylabel('Conversion Time (seconds)')

# Purchase amount comparison
sns.boxplot(data=ab_df, x='group', y='purchase_amount',
            palette='Set3', ax=axes[1])
axes[1].set_title('Purchase Amount by Test Group')
axes[1].set_xlabel('Test Group')
axes[1].set_ylabel('Purchase Amount ($)')

plt.suptitle('A/B Test Results: Checkout Flow Optimization', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Summary statistics
print("\nA/B Test Summary:")
print(ab_df.groupby('group')['conversion_time_sec'].agg(['mean', 'std']).round(1))
````

## 13. FacetGrid - Multi-Panel Conditional Plots

FacetGrid allows you to create a grid of plots conditioned on one or more categorical variables. This is extremely useful for comparing distributions or relationships across multiple subsets.

````python
# FacetGrid: bill distribution by day and time
g = sns.FacetGrid(tips_df, col='day', row='time', height=3.5, aspect=1.2)
g.map_dataframe(sns.histplot, x='total_bill', kde=True, color='teal')
g.set_axis_labels('Total Bill ($)', 'Count')
g.set_titles(col_template='{col_name}', row_template='{row_name}')
g.fig.suptitle('Bill Distribution by Day and Meal Time', y=1.02, fontsize=14)
plt.show()
````

## 15. Seaborn Color Palettes

Seaborn provides curated color palettes that are perceptually uniform and colorblind-friendly. This is critical for enterprise reports where accessibility matters.

````python
# Display common seaborn palettes
palettes = ['deep', 'muted', 'pastel', 'bright', 'dark', 'colorblind']

fig, axes = plt.subplots(len(palettes), 1, figsize=(10, len(palettes) * 0.8))
for ax, pal in zip(axes, palettes):
    colors = sns.color_palette(pal)
    for i, color in enumerate(colors):
        ax.barh(0, 1, left=i, color=color, height=0.6)
    ax.set_xlim(0, len(colors))
    ax.set_yticks([0])
    ax.set_yticklabels([pal], fontsize=11, fontweight='bold')
    ax.set_xticks([])
    ax.spines[:].set_visible(False)

plt.suptitle('Seaborn Color Palettes', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````
