```python
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

## 1. Vectors

A **vector** has both magnitude and direction. In NumPy we represent vectors as 1-D arrays.

### 1.1 Vector Addition and Scalar Multiplication

Vector addition adds corresponding components; scalar multiplication scales every component.

```python
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

# Vector addition
print('u + v =', u + v)

# Scalar multiplication
k = 2
print('k * u =', k * u)
```

### 1.2 Dot Product

The dot product sums the products of corresponding components:

$$\boldsymbol{u} \cdot \boldsymbol{v} = \sum_{i=1}^{n} u_i v_i = |\boldsymbol{u}||\boldsymbol{v}|\cos\theta$$

The geometric interpretation: dot product equals the product of the two vector magnitudes times the cosine of the angle between them.

```python
u = np.array([5, 1, 3])
m1 = np.array([4, 5, 1])
m2 = np.array([5, 1, 5])

# Dot products
print('u . m1 =', np.dot(u, m1))
print('u . m2 =', np.dot(u, m2))
```

### 1.3 Cosine Similarity

Cosine similarity measures how similar two vectors are by computing the cosine of the angle between them. A value close to 1 means high similarity.

$$\cos\theta = \frac{\boldsymbol{u} \cdot \boldsymbol{v}}{|\boldsymbol{u}||\boldsymbol{v}|}$$

**Example**: Recommend a movie to a user based on preference vectors.

```python
# User preference: [comedy, romance, action]  (1-5 scale)
user = np.array([5, 1, 3])

# Movie 1: romantic comedy  Movie 2: comedy action
movie1 = np.array([4, 5, 1])
movie2 = np.array([5, 1, 5])

cos1 = np.dot(user, movie1) / (np.linalg.norm(user) * np.linalg.norm(movie1))
cos2 = np.dot(user, movie2) / (np.linalg.norm(user) * np.linalg.norm(movie2))

print(f'Cosine similarity(user, movie1) = {cos1:.4f}')
print(f'Cosine similarity(user, movie2) = {cos2:.4f}')
print(f'\nRecommend movie {"2" if cos2 > cos1 else "1"} (higher similarity)')
```

### 1.4 Cross Product

The cross product of two 3-D vectors produces a new vector perpendicular to both:

$$\boldsymbol{A} \times \boldsymbol{B} = - (\boldsymbol{B} \times \boldsymbol{A})$$

Note: cross product is **anti-commutative**.

```python
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])

print('A x B =', np.cross(A, B))
print('B x A =', np.cross(B, A))
print('Anti-commutative check:', np.array_equal(np.cross(A, B), -np.cross(B, A)))
```

## 2. Matrices and Linear Algebra

A **matrix** is a rectangular array of numbers. NumPy supports matrix operations via 2-D arrays and the `numpy.linalg` module.

### 2.1 Creating Matrices

```python
# Using np.matrix (legacy, prefer 2-D ndarray)
m1 = np.matrix('1 2 3; 4 5 6')
print('m1 (matrix):')
print(m1)

# Using 2-D ndarray (recommended)
m2 = np.array([[1, 1], [2, 2], [3, 3]])
print('\nm2 (ndarray):')
print(m2)
```

### 2.2 Matrix Multiplication

- `np.matrix` uses `*` for matrix multiplication.
- For `ndarray`, use `@` operator or `np.matmul`.

Matrix multiplication is **not commutative**: generally AB != BA.

```python
A = np.array([[1, 0, 2], [-1, 3, 1]])   # 2x3
B = np.array([[3, 1], [2, 1], [1, 0]])   # 3x2

# Matrix multiplication with @ operator
C = A @ B
print('A @ B =')
print(C)

# Matrix multiplication with matmul
print('\nnp.matmul(A, B) =')
print(np.matmul(A, B))
```

### 2.3 Matrix Properties

| Property | Description |
|----------|-------------|
| `.T` | Transpose |
| `.shape` | Dimensions |
| `np.trace()` | Sum of diagonal elements |
| `np.linalg.det()` | Determinant |
| `np.linalg.matrix_rank()` | Rank |

```python
M = np.array([[1, 3, 5], [2, 4, 6], [4, 7, 9]])

print('Matrix M:')
print(M)
print(f'\nTranspose:\n{M.T}')
print(f'\nTrace = {np.trace(M)}')
print(f'Determinant = {np.linalg.det(M)}')
print(f'Rank = {np.linalg.matrix_rank(M)}')
```

### 2.4 Inverse Matrix

For a square matrix A, its inverse A^{-1} satisfies: A @ A^{-1} = I (identity matrix).

```python
A = np.array([[1., 2.], [3., 4.]])
A_inv = np.linalg.inv(A)

print('A =')
print(A)
print(f'\nA_inv =')
print(A_inv)
print(f'\nA @ A_inv (rounded) =')
print(np.around(A @ A_inv))
```

### 2.5 Eigenvalues and Eigenvectors

For matrix A, if Av = lambda*v, then lambda is an eigenvalue and v is the corresponding eigenvector.

```python
A = np.array([[4, -2], [1, 1]])
eigenvalues, eigenvectors = np.linalg.eig(A)

print('Matrix A:')
print(A)
print(f'\nEigenvalues: {eigenvalues}')
print(f'Eigenvectors (columns):\n{eigenvectors}')

# Verify: A @ v = lambda * v
for i in range(len(eigenvalues)):
    v = eigenvectors[:, i]
    lam = eigenvalues[i]
    print(f'\nVerify eigenvalue {lam}: A@v = {A @ v}, lambda*v = {lam * v}')
```

### 2.6 Solving Linear Equations

Given the system Ax = b:

$$\begin{cases}
x_1 + 2x_2 + x_3 = 8 \\
3x_1 + 7x_2 + 2x_3 = 23 \\
2x_1 + 2x_2 + x_3 = 9
\end{cases}$$

Use `np.linalg.solve` to find x.

```python
A = np.array([[1, 2, 1], [3, 7, 2], [2, 2, 1]])
b = np.array([8, 23, 9]).reshape(-1, 1)

# Check that the system has a unique solution
print(f'Rank of A: {np.linalg.matrix_rank(A)}')
print(f'Rank of [A|b]: {np.linalg.matrix_rank(np.hstack((A, b)))}')

# Method 1: linalg.solve
x = np.linalg.solve(A, b)
print(f'\nSolution (solve):\n{x}')

# Method 2: A^{-1} @ b
x2 = np.linalg.inv(A) @ b
print(f'Solution (inv @ b):\n{x2}')
```

### 2.7 Matrix Decompositions

NumPy supports several matrix decompositions useful in scientific computing.

#### QR Decomposition

```python
A = np.array([[1., 2.], [3., 4.], [5., 6.]])
Q, R = np.linalg.qr(A)

print('Q (orthogonal):')
print(Q)
print('\nR (upper triangular):')
print(R)
print(f'\nQ @ R (reconstructed):\n{np.around(Q @ R, decimals=10)}')
```

#### Singular Value Decomposition (SVD)

```python
A = np.array([[1., 2.], [3., 4.], [5., 6.]])
U, S, Vt = np.linalg.svd(A)

print('U:')
print(U)
print(f'\nSingular values: {S}')
print(f'\nV^T:')
print(Vt)
```

## 3. Polynomials

NumPy provides polynomial support via `np.poly1d` (legacy) and `numpy.polynomial.Polynomial` (recommended).

### 3.1 Creating and Operating on Polynomials

For f(x) = 3x^2 + 2x + 1:

```python
# Legacy poly1d (coefficients in descending order)
p1 = np.poly1d([3, 2, 1])
p2 = np.poly1d([1, 2, 3])

print('p1 =', p1)
print('p2 =', p2)
print(f'\np1 + p2 =\n{p1 + p2}')
print(f'\np1 * p2 =\n{p1 * p2}')
print(f'\np1(3) = {p1(3)}')
print(f'p2(3) = {p2(3)}')
```

### 3.2 Derivatives and Integrals

```python
p = np.poly1d([3, 2, 1])  # 3x^2 + 2x + 1

print(f'p(x) = {p}')
print(f'\nDerivative: {p.deriv()}')        # 6x + 2
print(f'\nIndefinite integral:\n{p.integ()}')  # x^3 + x^2 + x
```

### 3.3 Finding Roots

```python
p = np.poly1d([1, 3, 2])  # x^2 + 3x + 2 = (x+1)(x+2)
print(f'p(x) = {p}')
print(f'Roots: {p.roots}')
```

### 3.4 Modern Polynomial API (`numpy.polynomial.Polynomial`)

Note: coefficients are in **ascending** order (constant term first).

```python
from numpy.polynomial import Polynomial

# f(x) = 2 + 3x + x^2
p = Polynomial((2, 3, 1))

print(f'p(x) = {p}')
print(f'p(3) = {p(3)}')
print(f'Roots: {p.roots()}')
print(f'Degree: {p.degree()}')
print(f'Derivative: {p.deriv()}')
print(f'Integral: {p.integ()}')
```

## 4. Least-Squares Fitting

Polynomial least-squares fitting finds the coefficients that minimize the sum of squared residuals:

$$RSS = \sum_{i=0}^{k}(f(x_i) - y_i)^2$$

### 4.1 Example: Predict Online Shopping Spend from Income

```python
# Historical data: monthly income vs online shopping expenditure
x = np.array([
    25000, 15850, 15500, 20500, 22000, 20010, 26050, 12500, 18500, 27300,
    15000,  8300, 23320,  5250,  5800,  9100,  4800, 16000, 28500, 32000,
    31300, 10800,  6750,  6020, 13300, 30020,  3200, 17300,  8835,  3500
])
y = np.array([
    2599, 1400, 1120, 2560, 1900, 1200, 2320,  800, 1650, 2200,
     980,  580, 1885,  600,  400,  800,  420, 1380, 1980, 3999,
    3800,  725,  520,  420, 1200, 4020,  350, 1500,  560,  500
])

print(f'Number of samples: {len(x)}')
```

### 4.2 Scatter Plot

```python
plt.figure(figsize=(8, 5), dpi=120)
plt.scatter(x, y, color='steelblue', alpha=0.7, edgecolors='navy')
plt.xlabel('Monthly Income')
plt.ylabel('Online Shopping Expenditure')
plt.title('Income vs Online Shopping Spend')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 4.3 Correlation Coefficient

```python
corr_matrix = np.corrcoef(x, y)
print('Correlation coefficient matrix:')
print(corr_matrix)
print(f'\nPearson r = {corr_matrix[0, 1]:.4f}')
print('(Strong positive correlation close to 1.0)')
```

### 4.4 Linear Regression with `Polynomial.fit`

```python
from numpy.polynomial import Polynomial

# Fit a degree-1 polynomial (linear regression): y = ax + b
coeffs = Polynomial.fit(x, y, deg=1).convert().coef
print(f'Coefficients [b, a]: {coeffs}')
print(f'\nRegression equation: y = {coeffs[1]:.6f} * x + {coeffs[0]:.2f}')
```

### 4.5 Visualize the Regression Line

```python
a, b = coeffs[1], coeffs[0]
y_pred = a * x + b

plt.figure(figsize=(8, 5), dpi=120)
plt.scatter(x, y, color='steelblue', alpha=0.7, edgecolors='navy', label='Actual data')
plt.scatter(x, y_pred, color='tomato', alpha=0.7, edgecolors='darkred', label='Predicted')
# Sort for a clean line
sort_idx = np.argsort(x)
plt.plot(x[sort_idx], y_pred[sort_idx], color='darkcyan', linewidth=2, label='Regression line')
plt.xlabel('Monthly Income')
plt.ylabel('Online Shopping Expenditure')
plt.title('Least-Squares Linear Regression')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 4.6 Alternative: `np.polyfit`

```python
# polyfit returns coefficients in descending order: [a, b] for ax + b
coeffs_polyfit = np.polyfit(x, y, deg=1)
print(f'np.polyfit coefficients [a, b]: {coeffs_polyfit}')
print(f'Same result as Polynomial.fit: a={coeffs_polyfit[0]:.6f}, b={coeffs_polyfit[1]:.2f}')
```

## 5. Performance Tips

### 5.1 Vectorization vs Loops

NumPy vectorized operations are dramatically faster than Python loops because they delegate to optimized C/Fortran routines.

```python
import time

n = 1_000_000
a = np.random.randn(n)
b = np.random.randn(n)

# Python loop
start = time.perf_counter()
dot_loop = sum(ai * bi for ai, bi in zip(a, b))
time_loop = time.perf_counter() - start

# NumPy vectorized
start = time.perf_counter()
dot_np = np.dot(a, b)
time_np = time.perf_counter() - start

print(f'Loop result:    {dot_loop:.6f}  ({time_loop:.4f}s)')
print(f'NumPy result:   {dot_np:.6f}  ({time_np:.6f}s)')
print(f'\nNumPy is {time_loop / time_np:.0f}x faster')
```

### 5.2 Broadcasting vs Explicit Loops

```python
# Compute pairwise distances using broadcasting
n_points = 1000
points = np.random.randn(n_points, 2)

# Broadcasting approach
start = time.perf_counter()
diff = points[:, np.newaxis, :] - points[np.newaxis, :, :]  # (n, n, 2)
dist_broadcast = np.sqrt(np.sum(diff**2, axis=-1))           # (n, n)
time_broadcast = time.perf_counter() - start

print(f'Broadcasting: {dist_broadcast.shape} matrix in {time_broadcast:.4f}s')
print(f'Sample distances [0,1]={dist_broadcast[0,1]:.4f}, [0,2]={dist_broadcast[0,2]:.4f}')
```

### 5.3 Memory Layout: C-order vs F-order

Understanding memory layout can significantly impact performance for large arrays.

```python
rows, cols = 10000, 1000

# C-order (row-major, default)
arr_c = np.ascontiguousarray(np.random.randn(rows, cols))
# F-order (column-major)
arr_f = np.asfortranarray(arr_c.copy())

# Row-wise sum (favors C-order)
start = time.perf_counter()
_ = arr_c.sum(axis=1)
t_c_row = time.perf_counter() - start

start = time.perf_counter()
_ = arr_f.sum(axis=1)
t_f_row = time.perf_counter() - start

# Column-wise sum (favors F-order)
start = time.perf_counter()
_ = arr_c.sum(axis=0)
t_c_col = time.perf_counter() - start

start = time.perf_counter()
_ = arr_f.sum(axis=0)
t_f_col = time.perf_counter() - start

print(f'Row-wise sum:   C-order={t_c_row:.5f}s  F-order={t_f_row:.5f}s')
print(f'Col-wise sum:   C-order={t_c_col:.5f}s  F-order={t_f_col:.5f}s')
print(f'\nC-order is faster for row operations; F-order for column operations.')
```

### 5.4 Avoiding Unnecessary Copies

Use views (slicing) instead of copies when possible.

```python
arr = np.arange(1_000_000)

# View (no data copy)
view = arr[::2]
print(f'View shares memory: {np.shares_memory(arr, view)}')

# Explicit copy
copy = arr[::2].copy()
print(f'Copy shares memory: {np.shares_memory(arr, copy)}')

# In-place operations avoid allocation
start = time.perf_counter()
arr *= 2          # in-place
t_inplace = time.perf_counter() - start

arr = np.arange(1_000_000)
start = time.perf_counter()
arr = arr * 2     # creates new array
t_new = time.perf_counter() - start

print(f'\nIn-place multiply: {t_inplace:.6f}s')
print(f'New array multiply: {t_new:.6f}s')
```

### 5.5 Using `np.einsum` for Complex Operations

Einstein summation provides a concise way to express multi-dimensional operations and can be faster than chained operations.

```python
A = np.random.randn(100, 200)
B = np.random.randn(200, 300)

# Standard matrix multiply
start = time.perf_counter()
for _ in range(100):
    C1 = A @ B
t_matmul = time.perf_counter() - start

# einsum
start = time.perf_counter()
for _ in range(100):
    C2 = np.einsum('ij,jk->ik', A, B)
t_einsum = time.perf_counter() - start

print(f'matmul:  {t_matmul:.4f}s')
print(f'einsum:  {t_einsum:.4f}s')
print(f'Results match: {np.allclose(C1, C2)}')
```
