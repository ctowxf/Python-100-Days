## 1. Bubble Chart

A bubble chart extends a scatter plot by encoding a **third variable** as bubble size (`s`) and
optionally a **fourth** as color (`c`). Below we plot monthly income vs. online-shopping spend,
with bubble size proportional to purchase frequency.

````python
income  = np.array([5550, 7500, 10500, 15000, 20000, 25000, 30000, 40000])
outcome = np.array([800, 1800, 1250, 2000, 1800, 2100, 2500, 3500])
nums    = np.array([5, 3, 10, 5, 12, 20, 8, 10])

fig, ax = plt.subplots(figsize=(8, 5))
scatter = ax.scatter(income, outcome, s=nums * 30, c=nums, cmap='Reds',
                     edgecolors='grey', alpha=0.8)
fig.colorbar(scatter, ax=ax, label='Purchase Count')
ax.set_xlabel('Monthly Income')
ax.set_ylabel('Online Shopping Spend')
ax.set_title('Bubble Chart: Income vs Spend vs Frequency')
plt.tight_layout()
plt.show()
````

## 2. Area (Stacked) Chart

An area chart (stacked line chart) fills the region under each line, making it easy to see
**part-to-whole** relationships over a continuous axis. The example below shows how a person
spends time across a typical week.

````python
days = np.arange(7)
sleeping = [7, 8, 6, 6, 7, 8, 10]
eating   = [2, 3, 2, 1, 2, 3, 2]
working  = [7, 8, 7, 8, 6, 2, 3]
playing  = [8, 5, 9, 9, 9, 11, 9]

fig, ax = plt.subplots(figsize=(8, 4))
ax.stackplot(days, sleeping, eating, working, playing,
             labels=['Sleep', 'Eat', 'Work', 'Play'],
             colors=['#4c72b0', '#55a868', '#c44e52', '#8172b3'],
             alpha=0.85)
ax.set_xticks(days)
ax.set_xticklabels(['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])
ax.set_ylabel('Hours')
ax.set_title('Weekly Time Allocation (Stacked Area Chart)')
ax.legend(loc='upper left', fontsize=9)
plt.tight_layout()
plt.show()
````

## 3. Radar (Spider) Chart

Radar charts map a line chart onto a **polar coordinate** system — ideal for comparing
multiple quantitative dimensions of two or more entities. The key trick is to **close the
polygon** by appending the first data point at the end.

````python
labels = np.array(['Speed', 'Power', 'Experience', 'Defense', 'Serve', 'Technique'])
player_a = np.array([93, 95, 98, 92, 96, 97])   # Ma Long
player_b = np.array([30, 40, 65, 80, 45, 60])   # Mizutani

angles = np.linspace(0, 2 * np.pi, len(labels), endpoint=False)
# Close the polygon
angles = np.append(angles, angles[0])
player_a = np.append(player_a, player_a[0])
player_b = np.append(player_b, player_b[0])

fig, ax = plt.subplots(figsize=(5, 5), subplot_kw=dict(projection='polar'))
ax.plot(angles, player_a, 'o-', linewidth=2, label='Player A', color='#c44e52')
ax.fill(angles, player_a, alpha=0.25, color='#c44e52')
ax.plot(angles, player_b, 's-', linewidth=2, label='Player B', color='#4c72b0')
ax.fill(angles, player_b, alpha=0.20, color='#4c72b0')
ax.set_thetagrids(angles[:-1] * 180 / np.pi, labels, fontsize=10)
ax.set_ylim(0, 100)
ax.set_title('Radar Chart: Player Comparison', y=1.08)
ax.legend(loc='upper right', bbox_to_anchor=(1.25, 1.1))
plt.tight_layout()
plt.show()
````

## 4. Rose (Windrose / Nightingale) Chart

A rose chart is a bar chart projected onto **polar coordinates**. Invented by Florence
Nightingale, it exaggerates differences because bar **area** grows with the square of the
radius — great for highlighting cyclical patterns (quarters, months, weekdays).

````python
np.random.seed(42)
group1 = np.random.randint(20, 50, 4)
group2 = np.random.randint(10, 60, 4)

categories = [f'A-Q{i}' for i in range(1, 5)] + [f'B-Q{i}' for i in range(1, 5)]
values = np.concatenate([group1, group2])
theta = np.linspace(0, 2 * np.pi, len(categories), endpoint=False)
width = 2 * np.pi / len(categories)
colors = plt.cm.Set3(np.linspace(0, 1, len(categories)))

fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(projection='polar'))
bars = ax.bar(theta, values, width=width, color=colors, edgecolor='white', linewidth=0.5, bottom=0)
ax.set_thetagrids(theta * 180 / np.pi, categories, fontsize=9)
ax.set_title('Rose Chart: Quarterly Comparison', y=1.08)
plt.tight_layout()
plt.show()
````

## 5. 3D Surface Plot

Matplotlib supports basic 3D plotting via `mpl_toolkits.mplot3d`. For interactive rotation
and zoom, switch to a Qt backend (`%matplotlib qt`) and install `PyQt6`.

````python
fig = plt.figure(figsize=(9, 5))
ax = fig.add_subplot(111, projection='3d')

x = np.arange(-2, 2, 0.1)
y = np.arange(-2, 2, 0.1)
x, y = np.meshgrid(x, y)
z = (1 - y**5 + x**5) * np.exp(-x**2 - y**2)

surf = ax.plot_surface(x, y, z, cmap='viridis', edgecolor='none', alpha=0.9)
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10, label='z value')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Surface Plot')
plt.tight_layout()
plt.show()
````

## 6. Customization Deep-Dive

Matplotlib exposes **hundreds** of knobs. Below we demonstrate the most impactful ones:

- `rcParams` for global defaults
- Tick locators and formatters
- Annotations and arrows
- Twin axes (dual y-axis)
- Inset zoom plots

````python
# --- 6a. Dual Y-axis (twinx) ---
np.random.seed(0)
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.exp(x / 3)

fig, ax1 = plt.subplots(figsize=(8, 4))
ax1.plot(x, y1, color='#4c72b0', linewidth=2, label='sin(x)')
ax1.set_xlabel('x')
ax1.set_ylabel('sin(x)', color='#4c72b0')
ax1.tick_params(axis='y', labelcolor='#4c72b0')

ax2 = ax1.twinx()  # share the same x-axis
ax2.plot(x, y2, color='#c44e52', linewidth=2, linestyle='--', label='exp(x/3)')
ax2.set_ylabel('exp(x/3)', color='#c44e52')
ax2.tick_params(axis='y', labelcolor='#c44e52')

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')
ax1.set_title('Dual Y-Axis Chart')
fig.tight_layout()
plt.show()
````

````python
# --- 6b. Annotation + Arrow ---
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x, y1, color='#4c72b0', linewidth=2)
peak_x = np.pi / 2
peak_y = 1.0
ax.annotate('Peak (pi/2, 1)', xy=(peak_x, peak_y),
            xytext=(peak_x + 1.5, peak_y - 0.3),
            fontsize=11, arrowprops=dict(arrowstyle='->', color='black'),
            bbox=dict(boxstyle='round,pad=0.3', facecolor='lightyellow'))
ax.set_title('Annotation Example')
fig.tight_layout()
plt.show()
````

````python
# --- 6c. Inset (Zoom) Plot ---
from mpl_toolkits.axes_grid1.inset_locator import inset_locator, mark_inset
from mpl_toolkits.axes_grid1.inset_locator import InsetPosition

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, np.sin(x) + 0.1 * np.random.randn(len(x)), color='#4c72b0', alpha=0.6, label='noisy sin')
ax.plot(x, np.sin(x), color='#c44e52', linewidth=2, label='sin(x)')

# Create inset axes
axins = ax.inset_axes([0.55, 0.05, 0.4, 0.45])
axins.plot(x, np.sin(x) + 0.1 * np.random.randn(len(x)), color='#4c72b0', alpha=0.6)
axins.plot(x, np.sin(x), color='#c44e52', linewidth=2)
x1, x2 = 1.0, 2.5
axins.set_xlim(x1, x2)
axins.set_ylim(np.sin(x1) - 0.2, np.sin(x2) + 0.2)
ax.indicate_inset_zoom(axins, edgecolor='black')
ax.set_title('Inset Zoom Plot')
ax.legend()
fig.tight_layout()
plt.show()
````

## 7. Style Sheets

Matplotlib ships with many built-in styles. Use `plt.style.available` to list them and
`plt.style.use(name)` to activate one globally. Below we compare several popular styles
side by side.

````python
print('Available styles:')
print(plt.style.available)
````

````python
# Compare a few styles on the same data
demo_styles = ['default', 'ggplot', 'seaborn-v0_8', 'fivethirtyeight', 'dark_background']
demo_x = np.linspace(0, 10, 80)
demo_y1 = np.sin(demo_x)
demo_y2 = np.cos(demo_x)

fig, axes = plt.subplots(1, len(demo_styles), figsize=(20, 3.5), sharey=True)
for ax, style_name in zip(axes, demo_styles):
    with plt.style.context(style_name):
        ax.plot(demo_x, demo_y1, label='sin')
        ax.plot(demo_x, demo_y2, label='cos')
        ax.set_title(style_name, fontsize=10)
        ax.legend(fontsize=7)
fig.suptitle('Matplotlib Style Comparison', fontsize=14, y=1.02)
fig.tight_layout()
plt.show()
````

````python
# Creating a custom style via rcParams
custom_rc = {
    'lines.linewidth': 2.5,
    'lines.linestyle': '-',
    'axes.facecolor': '#f7f7f7',
    'axes.grid': True,
    'grid.color': 'white',
    'grid.linewidth': 0.8,
    'font.size': 11,
    'figure.facecolor': 'white',
}

with plt.rc_context(custom_rc):
    fig, ax = plt.subplots(figsize=(6, 3.5))
    ax.bar(['Q1', 'Q2', 'Q3', 'Q4'], [45, 62, 58, 71], color=['#4c72b0', '#55a868', '#c44e52', '#8172b3'])
    ax.set_title('Custom Style via rcParams')
    ax.set_ylabel('Revenue ($k)')
    fig.tight_layout()
    plt.show()
````

## 8. C++ vs Python: Visualization Workflow Comparison

When working in C++, plotting is **not** built into the language. Here is a side-by-side
comparison of common C++ visualization approaches versus the Python/matplotlib equivalent.

````python
comparison_table = """
+--------------------------+-----------------------------------+--------------------------------------+
| Task                    | C++ Approach                      | Python (matplotlib)                  |
+--------------------------+-----------------------------------+--------------------------------------+
| Basic line plot         | gnuplot pipe / Matplotlib-cpp /   | plt.plot(x, y)                       |
|                         | Qt Charts / ROOT                  |                                      |
+--------------------------+-----------------------------------+--------------------------------------+
| Interactive rotation    | OpenGL + GLFW / VTK               | %matplotlib qt  (built-in)           |
| (3D)                    |                                   |                                      |
+--------------------------+-----------------------------------+--------------------------------------+
| Save figure to PNG      | stb_image / libpng + manual       | plt.savefig('out.png', dpi=150)      |
|                         | rasterization                     |                                      |
+--------------------------+-----------------------------------+--------------------------------------+
| Style / theme switch    | Manual color/font constants       | plt.style.use('ggplot')              |
+--------------------------+-----------------------------------+--------------------------------------+
| Data pipeline           | Read CSV with custom parser or    | np.loadtxt / pd.read_csv + plot      |
|                         | fast-csv library                  | in 2 lines                           |
+--------------------------+-----------------------------------+--------------------------------------+
| Dependency management   | vcpkg / conan (compile libs)      | pip install matplotlib               |
+--------------------------+-----------------------------------+--------------------------------------+
| Embed in application    | Qt widget / Dear ImGui            | mpl Qt backend / web (bokeh/plotly)  |
+--------------------------+-----------------------------------+--------------------------------------+
"""
print(comparison_table)
````

### C++ gnuplot Pipe Example (for reference)

```cpp
// plot_sin.cpp — compile: g++ -o plot_sin plot_sin.cpp
#include <cstdio>
#include <cmath>

int main() {
    FILE* gp = popen("gnuplot -persistent", "w");
    fprintf(gp, "set title 'sin(x) via gnuplot pipe'\n");
    fprintf(gp, "plot sin(x) with lines lw 2\n");
    fflush(gp);
    // Keep gnuplot window open
    getchar();
    pclose(gp);
    return 0;
}
```

The Python equivalent is simply:
```python
plt.plot(x, np.sin(x))
plt.title('sin(x)')
plt.show()
```

For heavy numerical C++ code that must stay in C++ but still needs publication-quality plots,
the **matplotlib-cpp** header-only library (`github.com/lava/matplotlib-cpp`) provides a thin
wrapper that calls the Python matplotlib C API under the hood.

## 9. Enterprise Dashboard Example

Below is a multi-panel dashboard that might appear in a **business intelligence** context:
quarterly KPIs, product revenue breakdown, customer satisfaction trend, and a regional
comparison rose chart — all in one figure.

````python
np.random.seed(7)

fig = plt.figure(figsize=(14, 10))
fig.suptitle('Enterprise Quarterly Dashboard', fontsize=16, fontweight='bold', y=0.98)

# ---- Panel 1: Revenue Bar Chart ----
ax1 = fig.add_subplot(2, 2, 1)
quarters = ['Q1', 'Q2', 'Q3', 'Q4']
revenue_2024 = [120, 145, 138, 162]
revenue_2025 = [130, 155, 148, 175]
x_pos = np.arange(len(quarters))
w = 0.35
ax1.bar(x_pos - w/2, revenue_2024, w, label='2024', color='#4c72b0')
ax1.bar(x_pos + w/2, revenue_2025, w, label='2025', color='#55a868')
ax1.set_xticks(x_pos)
ax1.set_xticklabels(quarters)
ax1.set_ylabel('Revenue ($M)')
ax1.set_title('Quarterly Revenue')
ax1.legend(fontsize=9)

# ---- Panel 2: Product Mix Pie ----
ax2 = fig.add_subplot(2, 2, 2)
products = ['SaaS', 'Hardware', 'Consulting', 'Support']
shares = [42, 28, 18, 12]
explode = (0.05, 0, 0, 0)
ax2.pie(shares, explode=explode, labels=products, autopct='%1.1f%%',
        colors=['#4c72b0', '#55a868', '#c44e52', '#8172b3'],
        startangle=140, textprops={'fontsize': 9})
ax2.set_title('Product Revenue Mix (2025)')

# ---- Panel 3: Customer Satisfaction Trend ----
ax3 = fig.add_subplot(2, 2, 3)
months = np.arange(1, 13)
csat = [4.1, 4.0, 4.2, 4.3, 4.5, 4.4, 4.6, 4.5, 4.7, 4.6, 4.8, 4.7]
ax3.plot(months, csat, marker='o', color='#c44e52', linewidth=2)
ax3.fill_between(months, csat, alpha=0.15, color='#c44e52')
ax3.set_xlabel('Month')
ax3.set_ylabel('CSAT Score (1-5)')
ax3.set_title('Customer Satisfaction Trend')
ax3.set_xticks(months)
ax3.set_ylim(3.8, 5.0)
ax3.axhline(y=4.5, color='grey', linestyle='--', linewidth=0.8, label='Target')
ax3.legend(fontsize=8)

# ---- Panel 4: Regional Rose Chart ----
ax4 = fig.add_subplot(2, 2, 4, projection='polar')
regions = ['NA', 'EU', 'APAC', 'LATAM', 'MEA']
region_rev = [45, 38, 52, 20, 15]
theta = np.linspace(0, 2 * np.pi, len(regions), endpoint=False)
width = 2 * np.pi / len(regions)
colors = plt.cm.Paired(np.linspace(0, 1, len(regions)))
ax4.bar(theta, region_rev, width=width, color=colors, edgecolor='white', linewidth=0.5)
ax4.set_thetagrids(theta * 180 / np.pi, regions, fontsize=9)
ax4.set_title('Regional Revenue ($M)', y=1.08, fontsize=10)

fig.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
````

## Key Takeaways

1. **Bubble / Area / Radar / Rose / 3D** charts each solve a specific multi-dimensional
   communication problem — choose the one that matches your data story.
2. **Customization** (rcParams, annotations, twin axes, insets) turns a basic chart into a
   publication-ready figure.
3. **Style sheets** provide instant visual polish; create your own for brand consistency.
4. In **C++**, plotting requires external libraries (gnuplot, Qt Charts, matplotlib-cpp, VTK);
   Python's matplotlib gives you the same output in far fewer lines.
5. For **enterprise dashboards**, combine multiple subplots in a single figure with consistent
   colors and clear titles.
