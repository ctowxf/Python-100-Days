## Part 1: Descriptive Statistics

Descriptive statistics summarize the main features of a dataset: central tendency, dispersion, and shape.

### 1.1 Measures of Central Tendency

Central tendency tells us where the center of the data lies: **mean**, **median**, and **quantiles**.

```python
data = np.array([46, 51, 15, 42, 53, 71, 20, 62, 6, 94])
print('Data:', data)
print()

# Mean (average)
print('Sum:      ', data.sum(), '==', np.sum(data))
print('Mean:     ', data.mean(), '==', np.mean(data))

# Median (middle value)
print('Median:   ', np.median(data))

# Quantiles (percentiles)
print('Q1 (25%): ', np.quantile(data, 0.25))
print('Q2 (50%): ', np.quantile(data, 0.50))
print('Q3 (75%): ', np.quantile(data, 0.75))
```

### 1.2 Measures of Dispersion

Dispersion tells us how spread out the data is: **min**, **max**, **range**, **variance**, **standard deviation**.

```python
# Min, Max, Range
print('Min:  ', data.min(), '==', np.amin(data))
print('Max:  ', data.max(), '==', np.amax(data))
print('Range:', np.ptp(data))  # peak-to-peak (max - min)

# IQR (Interquartile Range)
q1, q3 = np.quantile(data, [0.25, 0.75])
print('IQR:  ', q3 - q1)
print()

# Variance and Standard Deviation
print('Variance:     ', data.var(), '==', np.var(data))
print('Std Deviation:', data.std(), '==', np.std(data))

# Coefficient of Variation (CV)
print('CV:           ', data.std() / data.mean())
```

### 1.3 Statistics Along Axes (2D Arrays)

For multi-dimensional arrays, use the `axis` parameter to compute statistics along rows or columns.

```python
scores = np.array([[72, 64, 73],
                    [61, 73, 61],
                    [76, 85, 77],
                    [97, 88, 90],
                    [63, 93, 82]])
print('Score matrix (5 students x 3 subjects):')
print(scores)
print()

# Overall statistics
print('Overall mean: ', scores.mean())
print()

# Along axis=0 (collapse rows -> one value per column/subject)
print('Mean per subject (axis=0):', scores.mean(axis=0))
print('Max  per subject (axis=0):', scores.max(axis=0))
print()

# Along axis=1 (collapse columns -> one value per row/student)
print('Mean per student (axis=1):', scores.mean(axis=1))
print('Max  per student (axis=1):', scores.max(axis=1))
```

### 1.4 Advanced Statistics with scipy.stats

NumPy does not provide geometric mean, harmonic mean, skewness, or kurtosis. Use `scipy.stats` for these.

```python
from scipy import stats

print('Arithmetic mean:     ', np.mean(data))
print('Geometric mean:      ', stats.gmean(data))
print('Harmonic mean:       ', stats.hmean(data))
print('Trimmed mean [10,90]:', stats.tmean(data, [10, 90]))
print('Coeff. of variation: ', stats.variation(data))
print('Skewness:            ', stats.skew(data))
print('Kurtosis:            ', stats.kurtosis(data))
```

### 1.5 Cumulative and Aggregate Functions

```python
arr = np.array([3, 1, 4, 1, 5, 9, 2, 6])
print('Array:          ', arr)
print()

# Cumulative sum and product
print('Cumulative sum: ', np.cumsum(arr))
print('Cumulative prod:', np.cumprod(arr))
print()

# Sorting
print('Sorted:         ', np.sort(arr))
print('Sorted indices: ', np.argsort(arr))

# Unique values and counts
values, counts = np.unique(arr, return_counts=True)
print('Unique values:  ', values)
print('Counts:         ', counts)
```

## Part 2: Random Number Generation

The `numpy.random` module provides functions for generating random numbers from various distributions, essential for simulation, testing, and Monte Carlo methods.

### 2.1 Setting the Random Seed

Setting a seed ensures reproducibility of random results.

```python
np.random.seed(42)
print('Random int [0, 10):', np.random.randint(0, 10))

# Reset seed -- same sequence again
np.random.seed(42)
print('Random int [0, 10):', np.random.randint(0, 10))
```

### 2.2 Generating Random Integers

```python
np.random.seed(0)

# Single random integer in [low, high)
print('Single random int:', np.random.randint(1, 100))
print()

# 1D array of random integers
arr_1d = np.random.randint(1, 100, size=10)
print('1D array (10 ints):', arr_1d)
print()

# 2D array of random integers
arr_2d = np.random.randint(1, 100, size=(3, 4))
print('2D array (3x4 ints):')
print(arr_2d)
```

### 2.3 Generating Random Floats

```python
np.random.seed(42)

# Uniform distribution in [0.0, 1.0)
uniform_arr = np.random.rand(3, 4)
print('Uniform [0, 1) - 3x4:')
print(uniform_arr)
print()

# Uniform in a custom range [low, high)
custom_range = np.random.uniform(10, 20, size=5)
print('Uniform [10, 20) - 5 elements:', custom_range)
```

### 2.4 Normal (Gaussian) Distribution

```python
np.random.seed(42)

# Standard normal (mean=0, std=1)
std_normal = np.random.randn(3, 3)
print('Standard normal (3x3):')
print(std_normal)
print()

# Normal with custom mean and std
custom_normal = np.random.normal(loc=100, scale=15, size=10)
print('Normal (mean=100, std=15) - 10 samples:', custom_normal.round(2))
```

### 2.5 Random Choice and Permutations

```python
np.random.seed(42)

# Random choice from an array
colors = np.array(['red', 'green', 'blue', 'yellow', 'purple'])
print('Random choice:', np.random.choice(colors))
print('3 random choices:', np.random.choice(colors, size=3, replace=False))
print()

# Shuffle (in-place)
deck = np.arange(1, 11)
print('Before shuffle:', deck)
np.random.shuffle(deck)
print('After shuffle: ', deck)
print()

# Permutation (returns new array)
original = np.arange(1, 6)
print('Original:     ', original)
print('Permutation:  ', np.random.permutation(original))
print('Unchanged:    ', original)
```

### 2.6 Other Useful Distributions

```python
np.random.seed(42)

# Binomial distribution (n trials, p probability)
binom = np.random.binomial(n=10, p=0.5, size=10)
print('Binomial (n=10, p=0.5):', binom)

# Poisson distribution (lambda rate)
poisson = np.random.poisson(lam=3.0, size=10)
print('Poisson (lambda=3):    ', poisson)

# Exponential distribution
expon = np.random.exponential(scale=2.0, size=5)
print('Exponential (scale=2): ', expon.round(2))

# Discrete uniform (dice rolls)
du = np.random.randint(1, 7, size=10)
print('Dice rolls (1-6):      ', du)
```

### 2.7 Practical Example: Monte Carlo Estimation of Pi

```python
np.random.seed(42)

# Generate random points in [0, 1) x [0, 1)
n_points = 100000
x = np.random.rand(n_points)
y = np.random.rand(n_points)

# Check if point is inside the unit quarter-circle
inside = (x**2 + y**2) <= 1.0

# Pi estimate = 4 * (fraction of points inside)
pi_estimate = 4 * np.sum(inside) / n_points
print(f'Estimated Pi: {pi_estimate:.6f}')
print(f'Actual Pi:    {np.pi:.6f}')
print(f'Error:        {abs(pi_estimate - np.pi):.6f}')
```

## Part 3: File I/O

NumPy provides several ways to save and load array data: binary format (`.npy`, `.npz`) and text format (`.txt`, `.csv`).

### 3.1 Binary Format: save / load (.npy)

The simplest approach for a single array. Saves in NumPy's native binary format -- fast and lossless.

```python
arr = np.arange(1, 10).reshape(3, 3)
print('Original array:')
print(arr)
print()

# Save to .npy file
np.save('output/array.npy', arr)
print('Saved to output/array.npy')

# Load from .npy file
loaded = np.load('output/array.npy')
print('Loaded from file:')
print(loaded)
print('Arrays equal:', np.array_equal(arr, loaded))
```

### 3.2 Binary Format: savez / savez_compressed (.npz)

Save multiple arrays into a single `.npz` archive.

```python
a = np.array([1, 2, 3])
b = np.array([[4, 5], [6, 7]])
c = np.arange(10)

# Save multiple arrays (compressed)
np.savez_compressed('output/multi_arrays.npz', first=a, second=b, third=c)
print('Saved 3 arrays to output/multi_arrays.npz')

# Load and access by name
data = np.load('output/multi_arrays.npz')
print('Keys:', list(data.keys()))
print('first: ', data['first'])
print('second:')
print(data['second'])
print('third: ', data['third'])
```

### 3.3 Text Format: savetxt / loadtxt (.txt / .csv)

Save arrays as human-readable text files. Useful for interoperability with other tools.

```python
scores = np.array([[72, 64, 73],
                    [61, 73, 61],
                    [76, 85, 77],
                    [97, 88, 90],
                    [63, 93, 82]], dtype=float)

# Save as CSV with header
np.savetxt('output/scores.csv', scores,
           delimiter=',', fmt='%.1f',
           header='Math,Physics,English', comments='')
print('Saved scores.csv')

# Show file content
with open('output/scores.csv') as f:
    print(f.read())
```

```python
# Load CSV back
loaded_scores = np.loadtxt('output/scores.csv', delimiter=',', skiprows=1)
print('Loaded scores:')
print(loaded_scores)
print('Arrays equal:', np.array_equal(scores, loaded_scores))
```

### 3.4 Text Format: genfromtxt (handling missing data)

`genfromtxt` is more powerful than `loadtxt` -- it handles missing values and mixed data types.

```python
# Write a CSV file with missing values
with open('output/incomplete.csv', 'w') as f:
    f.write('Name,Score1,Score2\n')
    f.write('Alice,85,90\n')
    f.write('Bob,,78\n')
    f.write('Charlie,92,\n')
    f.write('Diana,88,95\n')

print('File with missing values:')
with open('output/incomplete.csv') as f:
    print(f.read())

# Load with genfromtxt - missing values become NaN
data = np.genfromtxt('output/incomplete.csv', delimiter=',',
                     skip_header=1, filling_values=-1)
print('Loaded (missing -> -1):')
print(data)
```

### 3.5 Array dump / tofile Methods

The `ndarray` object itself has methods for writing data to files.

```python
arr = np.array([46, 51, 15, 42, 53, 71, 20, 62, 6, 94])

# dump: save as a single Python object (can reload with np.load)
arr.dump('output/array_dump.dat')
loaded_dump = np.load('output/array_dump.dat', allow_pickle=True)
print('dump/load result:', loaded_dump)
print()

# tofile: write raw binary or text
arr.tofile('output/array_raw.txt', sep=',')
print('tofile content:')
with open('output/array_raw.txt') as f:
    print(f.read())

# Read back with fromfile
loaded_fromfile = np.fromfile('output/array_raw.txt', sep=',', dtype=int)
print('fromfile result:', loaded_fromfile)
```

### 3.6 Summary: When to Use Which Format

| Method | Format | Use Case |
|---|---|---|
| `save` / `load` | `.npy` (binary) | Single array, fast read/write |
| `savez` / `savez_compressed` | `.npz` (binary) | Multiple arrays in one file |
| `savetxt` / `loadtxt` | `.txt` / `.csv` | Human-readable, interoperable |
| `genfromtxt` | `.txt` / `.csv` | Text files with missing data |
| `dump` / `load` | binary | Quick single-array persistence |
| `tofile` / `fromfile` | binary or text | Raw data, low-level control |
